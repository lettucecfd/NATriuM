```python
import numpy as np
import matplotlib.pyplot as plt
sourceDir = "/home/philipp/NATriuM/NATriuM/src/examples/step-grid-in/mesh/varyRefinement"
```

```python
filename = sourceDir + "/naca0012_res100_linear.txt"
x = []
y = []
with open(filename, "r") as file:
  string = file.read()
  string = string.splitlines()[1:-1]
  for line in string:
    if line not in ['', ' ']:
      line = line.split(' ')
      line = [l for l in line if l != '']
      xi = float(line[0])
      yi = float(line[1])
      x.append(xi)
      y.append(yi)
xA = np.array(x)
yA = np.array(y)
xStepped = []
yStepped = []
for i in range(len(x)):
  xStepped.append(x[i])
  yStepped.append(y[i])
  if (i < len(x)-1 and abs(y[i+1]-y[i])>1e-3):
    xStepped.append(x[i])
    yStepped.append(y[i+1])
xLin = np.array(xStepped)
yLin = np.array(yStepped)
```

```python
coords = np.stack((xLin,yLin))
# coords = np.sort(coords,axis=0)
plt.plot(coords[0,:],coords[1,:])
plt.show()
print(coords.shape)

dx = np.sqrt(np.pow(xLin[1:] - xLin[:-1],2) + np.pow(yLin[1:] - yLin[:-1],2))
plt.plot(xLin[:-1],dx)
plt.show()
```

```python
xString = "x={"
for x in xLin[:-1]:
  xString += (str(x)+", ") 
xString += str(xLin[-1]) + "};"
yString = "y={"
for y in yLin[:-1]:
  yString += (str(y)+", ") 
yString += str(yLin[-1]) + "};"
print(xString)
print(yString)
```
